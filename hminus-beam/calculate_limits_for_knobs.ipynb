```python
import numpy as np
```

```python
# calculated matrix
# matrix is of the form:
ELENA_KNOBS_CORRECTORS_MATRIX = np.array([
# h_offset v_offset h_angle  v_angle
    [-71.53,   0.00,  122.09,    0.00 ], # H2
    [  0.00,   4.30,    0.00, -229.82 ], # V2
    [-38.81,   0.00, -144.67,    0.00 ], # H3
    [  0.00, 112.25,    0.00,   -5.47 ]  # V3
])
```

```python
# knobs are H_offset_mm V_offset_mm H_angle_urad V_angle_urad


knobs = np.array([1.0, 1.0, 1.0, 0.0])

ELENA_KNOBS_CORRECTORS_MATRIX @ knobs

correctors = np.array([-82.0,232.0,-743.0,685.0])

settings = np.zeros((41,41),bool)

for i,hangle_urad in enumerate(np.arange(-20,20,1)):
    for j,vangle_urad in enumerate(np.arange(-20,20,1)):
        knobs = np.array([-6.0,4.0,hangle_urad,vangle_urad])
        corrected = correctors + ELENA_KNOBS_CORRECTORS_MATRIX @ knobs
        if np.all(corrected < 1500) and np.all( corrected > -1500.0):
            settings[i,j] = True
            print(f"Valid option: h_angle={hangle_urad} v_angle={vangle_urad} {corrected}" )
```
